```python
from obspy.taup import TauPyModel

model = TauPyModel(model="ak135")

source_depth_in_km=100
distance_in_degree=120
phase_list=["P", "S", "Pdiff", "PKP", "PKJKP"]
```

```python
arrivals = model.get_travel_times(
    source_depth_in_km=source_depth_in_km,
    distance_in_degree=distance_in_degree,
    phase_list=phase_list
)

print(arrivals)
```

```python
import matplotlib.pyplot as plt
from obspy.taup.tau import plot_ray_paths

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

plot_ray_paths(
    source_depth=source_depth_in_km, # Source depth in km
    phase_list=phase_list,           # Target seismic phases
    min_degrees=distance_in_degree,  # Pin minimum distance to 35
    max_degrees=distance_in_degree,  # Pin maximum distance to 35
    ax=ax, 
    fig=fig, 
    legend=True
)

plt.show()
```

```python
import matplotlib.pyplot as plt
from obspy.taup.tau import plot_ray_paths

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

plot_ray_paths(
    source_depth=source_depth_in_km,
    phase_list=phase_list,
    npoints=36,
    ax=ax, 
    fig=fig, 
    legend=True
)

plt.show()
```
